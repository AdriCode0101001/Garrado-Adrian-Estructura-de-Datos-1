# Unidad 1: Tipos de Datos Abstractos (TDA) y Gestión de Memoria

## 1. Fundamentos Teóricos

### ¿Qué es un TDA?
Un **Tipo de Dato Abstracto** es un modelo matemático de un tipo de datos donde el tipo se define por su comportamiento (operaciones) desde el punto de vista del *usuario*, independientemente de su implementación.

### Gestión de Memoria
Para entender los TDAs, es crucial comprender cómo se gestionan los datos en memoria:
*   **Stack (Pila):** Gestión automática, de tamaño fijo y rápido acceso. Se usa para variables locales y llamadas a funciones.
*   **Heap (Montículo):** Gestión dinámica. Permite reservar espacio en tiempo de ejecución. Es aquí donde residen las estructuras de datos dinámicas.

## 2. Esquema Conceptual

```mermaid
graph TD
    A[Usuario / Cliente] -->|Usa interfaces| B(TDA)
    B -->|Encapsula| C[Datos]
    B -->|Define| D[Operaciones/Lógica]
    C -->|Almacenamiento| E{Memoria}
    E --> F[Estática / Stack]
    E --> G[Dinámica / Heap]
```

### Desafío 2: Implementación de TDA Pila (Stack)

````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        # ERROR: Lógica invertida
        return len(self.items) > 0

    def push(self, item):
        # ERROR: No está agregando el item recibido
        self.items = [item]

    def pop(self):
        # ERROR: No verifica si hay elementos antes de borrar
        # y no retorna el elemento eliminado
        self.items.remove(-1)

# --- Prueba del Alumno ---
mi_pila = Stack()
mi_pila.push("A")
mi_pila.push("B")

print("¿Está vacía?", mi_pila.is_empty())
print("Elemento sacado:", mi_pila.pop())

# TODO: Corregir la clase Stack para que cumpla con el comportamiento LIFO
````

### Desafío 3: Gestión de Memoria en el Heap (Nodos Dinámicos)

**Contexto:** En lenguajes de bajo nivel, el Heap requiere gestión manual. En Python es automático, pero conceptualmente, si perdemos la referencia al 'puntero' inicial, perdemos el acceso a los datos.

**Problema:** El siguiente código intenta crear una secuencia de dos nodos en el Heap, pero debido a una mala asignación, el segundo nodo reemplaza al primero en la variable de control, perdiendo la referencia al inicio de la estructura.

````python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None # Este es nuestro 'puntero' al siguiente espacio en el Heap

# --- Prueba del Alumno ---
# El estudiante intenta crear una cadena: [Nodo 1] -> [Nodo 2]

contenedor = Nodo("Datos Importantes 1")

# ERROR: El alumno intenta agregar el segundo nodo de esta forma:
contenedor.siguiente = Nodo("Datos Importantes 2")

# Verificación
print(f"Contenido actual: {contenedor.valor}")
if contenedor.siguiente is None:
    print("ERROR: Se ha perdido la referencia al primer nodo. ¡Memory Leak conceptual!")

# TODO para el alumno: Corregir el código para que 'contenedor' mantenga el acceso al Nodo 1
# y el Nodo 1 apunte al Nodo 2.
````

### Desafío 4: TDA Punto y Circunferencia (Paso por Referencia vs. Valor)

**Contexto:** Hemos definido un `Punto` y una `Circunferencia` que usa este `Punto` como su centro. En Python, los objetos se pasan por referencia. Esto puede llevar a comportamientos inesperados si no se maneja correctamente, simulando un problema donde se esperaría una copia por valor.

**Problema:** Un estudiante crea una circunferencia con un centro definido por un objeto `Punto`. Posteriormente, el estudiante modifica directamente las coordenadas del `Punto` original, esperando que la circunferencia mantenga su centro original. Sin embargo, debido a la naturaleza de las referencias, la circunferencia también cambia.

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Circunferencia:
    def __init__(self, centro, radio):
        self.centro = Punto(centro.x, centro.y)
        self.radio = radio

    def __str__(self):
        return f"Circunferencia con centro en {self.centro} y radio {self.radio}"

# --- Prueba del Alumno ---
mi_punto_original = Punto(1, 2)
mi_circunferencia = Circunferencia(mi_punto_original, 5)

print(f"Circunferencia inicial: {mi_circunferencia}")

# El alumno modifica el punto original
mi_punto_original.x = 10
mi_punto_original.y = 20


print(f"Circunferencia después de modificar el punto original: {mi_circunferencia}")

# TODO: Corregir la clase Circunferencia para que su centro no cambie si el Punto original es modificado.
# Es decir, la Circunferencia debe tener su propia copia de las coordenadas del centro.
````

### Desafío 5: TDA Automóvil y Garaje (Gestión de Objetos en Colecciones)

**Contexto:** Estamos modelando un `Automovil` y un `Garaje` que almacena una colección de automóviles. Cuando se trabaja con listas de objetos, es fácil cometer errores al añadir o modificar elementos, especialmente si no se comprende cómo Python maneja las referencias a objetos en el *heap*.

**Problema:** Un estudiante intenta añadir varios automóviles a un garaje. Sin embargo, debido a una reutilización incorrecta de una variable de objeto `Automovil` al añadir a la lista, termina con un garaje que parece contener el mismo automóvil repetido varias veces, en lugar de automóviles distintos.

````python
class Automovil:
    def __init__(self, marca, modelo, color):
        self.marca = marca
        self.modelo = modelo
        self.color = color

    def __str__(self):
        return f"{self.color} {self.marca} {self.modelo}"

class Garaje:
    def __init__(self):
        self.automoviles = []

    def agregar_automovil(self, auto):
        self.automoviles.append(auto)

    def mostrar_automoviles(self):
        print("Automóviles en el garaje:")
        for i, auto in enumerate(self.automoviles):
            print(f"  {i+1}. {auto}")

# --- Prueba del Alumno ---
mi_garaje = Garaje()

# El alumno intenta añadir dos coches diferentes
auto_temporal_1 = Automovil("Toyota", "Corolla", "Rojo")
mi_garaje.agregar_automovil(auto_temporal_1)

# ERROR: Reutiliza la misma variable para 'crear' otro coche, pero en realidad solo modifica el objeto existente
auto_temporal_2 = Automovil("Honda", "Civic", "Azul")
mi_garaje.agregar_automovil(auto_temporal_2)

mi_garaje.mostrar_automoviles()

# TODO: Corregir el código para que el garaje contenga dos automóviles distintos con sus respectivas propiedades.
````

### Desafío 6: TDA Línea (Inmutabilidad Conceptual con Objetos Mutables)

**Contexto:** Una `Línea` se define por dos `Punto`s: un punto de inicio y un punto final. Conceptualmente, una vez que una línea se define, sus puntos extremos no deberían cambiar si se modifican los `Punto`s originales que se usaron para crearla, a menos que se quiera explícitamente.

**Problema:** Similar al desafío de la circunferencia, si la clase `Línea` almacena referencias directas a los objetos `Punto` mutables pasados durante su inicialización, cualquier modificación externa a esos `Punto`s afectará la `Línea` de forma inesperada. El estudiante espera que la línea permanezca estática después de su creación.

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

class Linea:
    def __init__(self, inicio, fin):
        self.inicio = Punto(inicio.x, inicio.y)
        self.fin = Punto(fin.x, fin.y)
    def __str__(self):
        return f"Línea de {self.inicio} a {self.fin}"

# --- Prueba del Alumno ---
punto_a = Punto(0, 0)
punto_b = Punto(5, 5)

mi_linea = Linea(punto_a, punto_b)
print(f"Línea original: {mi_linea}")

# El alumno modifica uno de los puntos originales
punto_a.x = 10
punto_a.y = 10

print(f"Línea después de modificar el punto 'A': {mi_linea}")

# TODO: Corregir la clase Linea para que sea "inmutable" conceptualmente;
# es decir, que los cambios en los objetos Punto externos no afecten la línea una vez creada.
````
